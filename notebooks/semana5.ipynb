## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f1 = lambda x: np.e**x - x - 2
g1 = lambda x: np.e**x - 2


def main():
    r = pontofixo(-1.8, g1)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(-1.8, f1, df=lambda x: np.e**x - 1)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(-1.8, f1)
    print(f"raiz newton-raphson = {r}")
    r = secante(-1.8, -1.7, f1)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import numpy as np
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import pontofixo, newton_raphson, secante

f2 = lambda x: np.cos(x) - x**2
g2 = lambda x: x + f2(x) / (np.sin(x) + 2 * x)
df2 = lambda x: -np.sin(x) - 2 * x

raiz_ponto_fixo = pontofixo(1.0, g2, TOL=1e-8)
raiz_newton = newton_raphson(1.0, f2, TOL=1e-8, df=df2)
raiz_secante = secante(0.8, 1.0, f2, TOL=1e-8)

print(f'Ponto fixo:       {raiz_ponto_fixo:.8f}')
print(f'Newton-Raphson:   {raiz_newton:.8f}')
print(f'Secante:          {raiz_secante:.8f}')
print(f'Resíduo:          {abs(f2(raiz_ponto_fixo)):.2e}')
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import pontofixo, newton_raphson, secante

f3 = lambda x: np.exp(-x**2) - 2 * x
g3 = lambda x: np.exp(-x**2) / 2
df3 = lambda x: -2 * x * np.exp(-x**2) - 2

raiz_ponto_fixo = pontofixo(0.5, g3, TOL=1e-8)
raiz_newton = newton_raphson(0.5, f3, TOL=1e-8, df=df3)
raiz_secante = secante(0.4, 0.5, f3, TOL=1e-8)

print(f'Ponto fixo:       {raiz_ponto_fixo:.8f}')
print(f'Newton-Raphson:   {raiz_newton:.8f}')
print(f'Secante:          {raiz_secante:.8f}')
print(f'Resíduo:          {abs(f3(raiz_ponto_fixo)):.2e}')
````

## Atividade 4
Resolva os três últimos exercícios da semana anterior usando os métodos do ponto fixo.

````python
import numpy as np
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import pontofixo, newton_raphson, secante

TOL = 1e-10

# Exercicio 4: V = R * Id + vd, Id = IR * (exp(vd / vt) - 1)
k = 1.380649e-23
q = 1.602176634e-19
T = 300
IR = 1e-12
vt = k * T / q

casos_diodo = [
    (30, 1e3),
    (3, 1e3),
    (3, 10e3),
    (0.3, 1e3),
    (-0.3, 1e3),
    (-30, 1e3),
    (-30, 10e3),
]

print('Exercicio 4 - tensao no diodo (V)')
for V, R in casos_diodo:
    f4 = lambda vd, V=V, R=R: R * IR * np.expm1(vd / vt) + vd - V
    df4 = lambda vd, R=R: R * IR * np.exp(vd / vt) / vt + 1
    g4 = lambda vd, f=f4, df=df4: vd - f(vd) / df(vd)
    inicial = min(V, 0.7) if V > 0 else V
    raiz_ponto_fixo = pontofixo(inicial, g4, TOL)
    raiz_newton = newton_raphson(inicial, f4, TOL, df4)
    raiz_secante = secante(inicial, inicial + 0.1, f4, TOL)
    print(f'V={V:g} V, R={R / 1e3:g} kOhm: '
          f'PF={raiz_ponto_fixo:.6f}, '
          f'Newton={raiz_newton:.6f}, Secante={raiz_secante:.6f}')

# Exercicio 5: f = C * (cosh(d / (2C)) - 1)
d = 500
fmax = 50
f5 = lambda C: C * (np.cosh(d / (2 * C)) - 1) - fmax
df5 = lambda C: np.cosh(d / (2 * C)) - 1 - (d / (2 * C)) * np.sinh(d / (2 * C))
g5 = lambda C: C - f5(C) / df5(C)

raiz_ponto_fixo = pontofixo(600.0, g5, TOL)
raiz_newton = newton_raphson(600.0, f5, TOL, df5)
raiz_secante = secante(500.0, 800.0, f5, TOL)

print('\nExercicio 5 - comprimento do cabo (m)')
print(f'Ponto fixo:     {raiz_ponto_fixo:.4f}')
print(f'Newton-Raphson: {raiz_newton:.4f}')
print(f'Secante:        {raiz_secante:.4f}')

# Exercicio 6: Id = sin(beta - phi) + sin(phi) * exp(-beta / tan(phi))
f = 1e3
L = 100e-3
R = 1e3
phi = np.arctan(2 * np.pi * f * L / R)
f6 = lambda beta: np.sin(beta - phi) + np.sin(phi) * np.exp(-beta / np.tan(phi))
df6 = lambda beta: np.cos(beta - phi) - (np.sin(phi) / np.tan(phi)) * np.exp(-beta / np.tan(phi))
g6 = lambda beta: beta - f6(beta) / df6(beta)

raiz_ponto_fixo = pontofixo(3.7, g6, TOL)
raiz_newton = newton_raphson(3.7, f6, TOL, df6)
raiz_secante = secante(3.5, 3.9, f6, TOL)

print('\nExercicio 6 - angulo beta')
print(f'phi = {np.degrees(phi):.6f} graus')
print(f'Ponto fixo:     {np.degrees(raiz_ponto_fixo):.4f} graus')
print(f'Newton-Raphson: {np.degrees(raiz_newton):.4f} graus')
print(f'Secante:        {np.degrees(raiz_secante):.4f} graus')
````
